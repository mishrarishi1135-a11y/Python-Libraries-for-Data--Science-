```python
# Understand this problem
# you have data of some student which are in 3rd sem and columns(hours of study in sem, GPA)
# you see as soon as hours increase marks will also increase.
# Linear Regression: find the formula or relation between two quantity for giving the not prediction also approx answer.(with minimum error)
```

```python
# In seaborn, Regplot not create the scatter plot also create a linear regression line.
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
tips = sns.load_dataset("tips")
tips
```

```python
# Regression plot: regplot and lmplot
# both function draw the scatter plot of two variables, x and y and then fit the regression model y~x and plot the resulting regression line and a 95% confidence interval for that regression.
sns.scatterplot(data=tips, x='total_bill', y='tip')
```

```python
# But we want a regression line also
sns.regplot(data=tips, x='total_bill', y='tip')
# lighter shade on the both side of line --> error of line
```

```python
sns.lmplot(data=tips, x='total_bill', y='tip')  # when we use lmplot then we add a 'hue' parameter.
```

```python
sns.lmplot(data=tips, x='total_bill', y='tip', hue='sex')
# Hue paramater is not available in regplot.
```

```python
# Residual plot
sns.residplot(data=tips, x='total_bill', y='tip')
# linear regression model do some error that is known as residual.
# It's a error of model at tip prediction time.
```

```python
# Matrix plot --> Heatmap and cluster map
# Heatmap --> plot rectangular as a color-encoded matrix
import numpy as np
import seaborn as sns
import pandas as pd
import plotly.express as px
import matplotlib.pyplot as plt
gap =px.data.gapminder()
gap
```

```python
# plot "every country life expectation in every year"(also create some bins)
# grid -> no. of countries* no.of years
# It's a long dataset then we convert into the wide dataset.(using pivot function)
gap.pivot(index='country',columns='year',values='lifeExp')
```

```python
# Represent this complete grid into a heatmap
temp_df = gap.pivot(index='country',columns='year',values='lifeExp')
# Also heatmap is a axis level function and no any other method to plot heatmap.
sns.heatmap(temp_df)
```

```python
# according to your purpose you can reshape this
temp_df = gap.pivot(index='country',columns='year',values='lifeExp')
plt.figure(figsize=(15,10))
sns.heatmap(temp_df)
```

```python
temp_df = gap[gap['continent'] == 'Europe'].pivot(index='country', columns='year',values='lifeExp')
plt.figure(figsize=(15,15))
sns.heatmap(temp_df)
```

```python
# If you want to print the number or values on the graph line so you just use a parameter "annot"
temp_df = gap[gap['continent'] == 'Europe'].pivot(index='country', columns='year',values='lifeExp')
plt.figure(figsize=(15,15))
sns.heatmap(temp_df,annot=True)
```

```python
# For better readability you create a gap between the lines by using 'linewidth' parameter
temp_df = gap[gap['continent'] == 'Europe'].pivot(index='country', columns='year',values='lifeExp')
plt.figure(figsize=(15,15))
sns.heatmap(temp_df,annot=True,linewidth=0.5)
# You also use the c-maps for changing the color of graph like summer, viridis, rainy or jet.
```

```python
# Cluster map --> plot a matrix dataset as a hierarchically-clustered heatmap.
# This function requires scipy to be available.
# similar columns ko ek sath aur dis-similar column ko alag-alag
iris = px.data.iris()
iris
```

```python
sns.clustermap(iris.iloc[:,[0,1,2,3]])   # [:,[0,1,2,3]] --> sari rows aur 0,1,2,3 column
# first of all, it changes the columns order. arrange two columns according to similar data.
```

```python
tips = sns.load_dataset("tips")
sns.displot(data=tips, x='total_bill', kind='kde', hue='sex',fill=True)
```

```python
# How to change size in figure level plot.
# we do not use this "plt.figure(figsize=(15,7))" because figure level function have it's own figure
sns.displot(data=tips,x='total_bill',kind='kde',hue='sex',fill=True,height=10,aspect=2)
# use this parameter --> height for length and aspect for width.
```

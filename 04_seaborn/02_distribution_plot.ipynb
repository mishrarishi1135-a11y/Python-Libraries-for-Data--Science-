```python
# Distribution Plot
# used for univariate analysis, used to find out the distribution, range of the observation, central tendency
# also for check is the data bimodal? , are there outliers
# plots under distribution plot --> histplot, kdeplot, rugplot
# These plots tells us the how the data are distributed.
```

```python
# plotting univariate histogram
# here, figure level -> displot and axis level ->histplot, kdeplot, rugplot
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
sns.load_dataset("tips")
```

```python
sns.load_dataset("tips")
```

```python
tips = sns.load_dataset('tips')
sns.histplot(data=tips,x='total_bill')
```

```python
sns.displot(data=tips, x='total_bill', kind='hist')  # figure level function
```

```python
# bins (division part) parameter
sns.displot(data=tips, x='total_bill',kind='hist', bins=20)
```

```python
# It's also possible to visualize the distribution of a categorical variable using the logic of histogram.
# Discrete bins are automatically set for categorical variables.
sns.displot(data=tips, x='day',kind='hist')
# it is a count plot not histogram which tell the count of categorical column's categories.
```

```python
sns.displot(data=tips, x='tip', kind='hist',hue='sex')
```

```python
# If you want to see clear then use "element" parameter
sns.displot(data=tips, x='tip', kind='hist', hue='sex', element='step')
# creates only outline
```

```python
titanic = sns.load_dataset('titanic')
titanic
```

```python
# histogram of age
sns.displot(data=titanic, x='age', element='step',kind='hist')
```

```python
sns.displot(data=titanic, x='age', hue='sex',element='step',kind='hist')
```

```python
# we create two histogram foe male and female for better understanding
# faceting using col and row -> this function will not work on histplot function(axis level function).
# Faceting is possible with figure level function like displot.
sns.displot(data=tips, x='tip',kind='hist',col='sex',element='step')
```

```python
# 2-plot --> kernel density estimation(kde) plot
# rather than using discrete bins, a KDE plot smooths the observations with a gaussian kernel, producing a continuous density estimate
tips
```

```python
sns.kdeplot(data=tips, x='total_bill',)  # Roughly it says in y-axis have probability of total_bill.
# y-axis represent density of probability in your data
```

```python
# hue -> fill
sns.displot(data=tips, x='total_bill', kind='kde')
```

```python
# also use 'hue' parameter for better understanding
sns.displot(data=tips, x='total_bill', kind='kde', hue='sex')
```

```python
# provide also fill -> True
sns.displot(data=tips, x='total_bill', kind='kde', hue='sex',fill=True)
```

```python
# Rug plot -> plot marginal distribution by drawing ticks along the x and y axis.
# This function is intended to complement other plots by showing the location of individual observations in an unobtrusive way.
sns.kdeplot(data=tips, x='total_bill')
sns.rugplot(data=tips, x='total_bill')    # just another method to show information of same data.(orange=rug pot).
# Also rug plot represent the kind of density.
```

```python
# Bi variate histogram -> bins the data with in rectangles that tile the plot.
# and then shows the count of observations within each rectangle with the fill color.

# create a histogram of tip and total_bill together
sns.histplot(data=tips, x='total_bill', y='tip')
```

```python
sns.displot(data=tips, x='total_bill',y='tip', kind='hist')
```

```python
# Bi variate kde plot -> smoothes the (x,y) observations with a 2D gaussian
sns.kdeplot(data=tips, x='total_bill',y='tip')
```

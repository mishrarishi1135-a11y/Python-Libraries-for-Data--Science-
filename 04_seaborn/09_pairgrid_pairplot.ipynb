```python
# Plotting pairwise relationship(PairGrid vs PairPlot)
import seaborn as sns
import plotly.express as px
import matplotlib.pyplot as plt
iris = sns.load_dataset('iris')
iris
```

```python
# PairPlot detect automatically how many numerical column in data.
sns.pairplot(iris)    # only give the name of dataframe
# We have total (4*4=16 combinations) and it create histogram when x and y axis have same column name.
# all the diagrams which are above the diagonal and below the diagonal, both are same.
```

```python
# we add a hue parameter for better understanding
sns.pairplot(iris,hue='species')   # Now it create kde plot for the data of x and y axis.
```

```python
# Internally pair plot, pair grid ki help se ban ta hai.
# By using pair grid, we can perform some additional things.

# Pair grid
g = sns.PairGrid(data=iris)
# g.map
g.map(sns.scatterplot)       # all are create in scatter plot.
```

```python
# You can plot diagonal and non-diagonal items in different-2 space.
# map_diag -> map_offdiag
g = sns.PairGrid(iris,hue='species')
g.map_diag(sns.histplot)    # for diagonal items
g.map_offdiag(sns.scatterplot)  # for non-diagonal items

# best point of this, we plot easily any type of graph by using the pair grid.
# pair plot is the special case of pair grid.
# Also pair grid have no restriction for plotting any type of graph.
# You can plot different plot for above the diagonal and below the diagonal.
```

```python
# map_diag -> map_upper -> map_lower
g = sns.PairGrid(data=iris,hue='species')
g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.scatterplot)
```

```python
# You also work on a particular columns using 'vars' parameter.
g = sns.PairGrid(data=iris,hue='species',vars=['sepal_width','petal_width'])
g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.scatterplot)
```

```python
# Multi Grid plot
import seaborn as sns
import plotly.express as px
import matplotlib.pyplot as plt
# A second way to plot Facet plots -> FacetGrid
# we learn about these -> in figure level(relplot, displot, catplot, lmplot)
tips = sns.load_dataset("tips")
tips
```

```python
sns.catplot(data=tips, x='sex', y='total_bill', kind='violin')
```

```python
# But we have some extra feature like we add facet col on 'day'
sns.catplot(data=tips, x='sex', y='total_bill',kind='violin', col='day')
```

```python
# Now creating the facet plots using 'FacetGrid'
g = sns.FacetGrid(data=tips, col='day', row='time') # It create a object so we store it into a variable.
g.map(sns.violinplot,'sex','total_bill')
```

```python
g = sns.FacetGrid(data=tips, col='day', row='time') # It create a object so we store it into a variable.
g.map(sns.boxplot,'sex','total_bill')
```

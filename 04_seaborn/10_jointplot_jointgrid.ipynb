```python
import seaborn as sns
import plotly.express as px
import matplotlib.pyplot as plt
iris = sns.load_dataset('iris')
iris
```

```python
# Joint plot
tips = sns.load_dataset("tips")
tips
```

```python
sns.jointplot(data=tips,x='total_bill',y='tip')
# It also create a histogram for the particular x and y axis.
```

```python
# We have kind parameter like kde,hist,scatter,resid,hex and reg. (mix with 1D and 2D)
sns.jointplot(data=tips,x='total_bill',y='tip',kind='hist',hue='sex')
# You can get a combined view of 2D and 1D histogram with hue parameter.
```

```python
# Now understand joint grid --> here we can tell what first i want and second i want other.
g = sns.JointGrid(data=tips,x='total_bill',y='tip')
g.plot(sns.scatterplot,sns.histplot)
```

```python
# Now you have access to create any type of graph by using the grid function.
g = sns.JointGrid(data=tips,x='total_bill',y='tip')
g.plot(sns.scatterplot,sns.boxplot)
```

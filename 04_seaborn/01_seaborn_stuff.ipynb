```python
sns.load_dataset('tips')   # seaborn have multiple datasets.
```

```python
# we want to plot a scatter plot (generally on numerical quantities) --> It's a axis level function
import seaborn as sns
import matplotlib.pyplot as plt
tips = sns.load_dataset("tips")
sns.scatterplot(data=tips,x='total_bill', y='tip')
plt.show()
```

```python
# if you to figure level of this dataset(relplot --> figure level function --> square shape)
sns.relplot(data = tips, x='total_bill', y='tip',kind = 'scatter')

# generally we use figure level function much more than axis level function.
```

```python
# If we want to print as a human gender division --> using 'hue' parameter
sns.scatterplot(data=tips, x='total_bill',y='tip',hue='sex')
```

```python
# we can add much more information like about the time
sns.scatterplot(data=tips, x='total_bill',y='tip',hue='sex',style='time')
```

```python
# if we want to create the info about the size of customers
sns.scatterplot(data=tips, x='total_bill',y='tip',hue='sex',style='time',size='size')
```

```python
# Line plot --> to visualize the time series data
gap = px.data.gapminder()
gap
```

```python
temp_df = gap[gap['country'] == 'India']
temp_df
```

```python
# axis level function
sns.lineplot(data=temp_df, x='year', y='lifeExp')
```

```python
# using relplot  --> figure level function ( square shape)
sns.relplot(data = temp_df, x='year',y='lifeExp', kind='line')
```

```python
# hue --> style
temp_df = gap[gap['country'].isin(['India','Pakistan','China'])]
temp_df
```

```python
sns.relplot(kind='line', data=temp_df, x='year',y='lifeExp',hue='country')
# here rel plot me legend(info) value outside the graph.
```

```python
sns.lineplot(data=temp_df, x='year',y='lifeExp',hue='country') # axis level function
# line plot me info value graph ke inside hoti hai.
```

```python
temp_df = gap[gap['country'].isin(['India','Brazil','Germany'])]
temp_df
```

```python
sns.relplot(kind = 'line', data=temp_df, x='year', y='lifeExp', hue='country', style='continent')
```

```python
# Facet plot --> works only with figure level function --> means it will work with relplot
# it will not work with scatter plot function and line plot function
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter', hue='sex')
```

```python
# if we create two graph side by side one for male and one for women
# particular categorical column ke uper multiple plot bana skte hai --> Facet plot
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter', col='sex')
```

```python
# we create this graph on row basis
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter', row='sex')
```

```python
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter',col='smoker', row='sex')
# It create a 4 graph 2 for smoker and 2 for gender
# male smoker and male non-smoker, female smoker and female non-smoker
```

```python
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter',col='smoker',row='day')
```

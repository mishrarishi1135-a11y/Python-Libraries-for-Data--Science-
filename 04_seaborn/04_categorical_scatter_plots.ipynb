```python
# categorical plots
import numpy as np
import seaborn as sns
import pandas as pd
import plotly.express as px
import matplotlib.pyplot as plt
```

```python
# import datasets
tips = sns.load_dataset('tips')
```

```python
iris = sns.load_dataset('iris')
```

```python
sns.scatterplot(data=tips, x='total_bill', y='tip') # use in numerical data context
```

```python
# 1: Categorical Scatter plots --> strip plot and swarm plot
# Make sure one column is categorical and other column is numerical (for scatter plot on categorical data)

# Strip plot
tips
```

```python
# use two columns --> total_bill and day
sns.stripplot(data=tips, x='day', y='total_bill')
```

```python
# If you want all the points are arrange in one line then use this parameter 'jitter'
sns.stripplot(data=tips, x='day', y='total_bill',jitter=False)  # axis level function
```

```python
# same graph through figure level function --> catplot
sns.catplot(data=tips, x='day',y='total_bill',kind='strip')
```

```python
# Understand some changes in jitter parameter
sns.catplot(data=tips, x='day',y='total_bill',kind='strip', jitter=0.2)  # sare points expand ho jayenge
```

```python
# we also use 'hue' parameter
sns.catplot(data=tips, x='day', y='total_bill', kind='strip', hue='sex')
```

```python
# swarm plot
sns.catplot(data=tips, x='day', y='total_bill', kind='swarm') # better representation of distribution of data.
```

```python
sns.swarmplot(data=tips, x='day', y='total_bill', hue='sex')
# swarm plot is essential for small dataset.
```

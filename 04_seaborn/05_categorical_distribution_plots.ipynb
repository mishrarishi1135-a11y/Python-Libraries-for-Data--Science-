```python
# Categorical distribution plot --> boxplot and violin plot.
import seaborn as sns
import plotly.express as px
import matplotlib.pyplot as plt
# distribution plot tells the distribution of single column while scatter done bi-variate analysis.
```

```python
# Box plot --> It is standardized way of displaying the distribution of data based on five number summary.
# 5 number summary ('minimum',first quartile[Q1],median, third quartile[Q3], and 'maximum')
# Q1 represent the 25th percentile while Q3 represent the 75th percentile also median represent the Q2(50th) percentile.

# It can tell you about your outliers and what their values are.
# Also if your data is symmetrical, how tightly your data is grouped and if and how your data is skewed.
# "minimum" --> Q1 - 1.5 * IQR(inter quartile range) and "maximum" --> Q3 + 1.5 * IQR
```

```python
tips = sns.load_dataset('tips')
tips
```

```python
sns.boxplot(data=tips, x='sex', y='total_bill')
```

```python
sns.boxplot(data=tips, x='day', y='total_bill')
```

```python
# Box plot is axis level function so we plot figure level function using catplot
sns.catplot(data=tips, x='day', y='total_bill', kind='box')  # This will gave same result.
```

```python
# also you add "hue" parameter.
sns.catplot(data=tips, x='day', y='total_bill', kind='box',hue='sex')
```

```python
# If you want only total bill box plot, then single boxplot --> numerical column.
sns.boxplot(data=tips, y='total_bill')
```

```python
# Violin plot = (Box plot + KDE plot)
sns.violinplot(data=tips, x='day', y='total_bill')
# Inside the middle which are box plot and outside structure is kde plot.
```

```python
sns.catplot(data=tips, x='day', y='total_bill', kind='violin')
```

```python
# also add a 'hue' parameter
sns.catplot(data=tips, x='day', y='total_bill', kind='violin', hue='sex')
```

```python
# Much more better understanding use "split" parameter
sns.catplot(data=tips,x='day', y='total_bill', kind='violin',hue='sex',split=True)
```

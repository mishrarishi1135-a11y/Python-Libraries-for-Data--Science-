```python
# ============================================================
# BUSINESS DATA ANALYSIS - OCTOBER
# Libraries: Pandas, Matplotlib, Seaborn
# Output: PNG graphs, CSV data, ZIP file
# ============================================================

import os
import shutil
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ------------------------------------------------------------
# 1. SETUP
# ------------------------------------------------------------

sns.set_theme(style="whitegrid")

BASE_DIR = os.getcwd()
OUTPUT_DIR = os.path.join(BASE_DIR, "outputs")
GRAPH_DIR = os.path.join(OUTPUT_DIR, "graphs")

os.makedirs(GRAPH_DIR, exist_ok=True)

# Start graph numbering from 1
graph_counter = 0

print("=" * 60)
print("       OCTOBER BUSINESS DATA ANALYSIS")
print("=" * 60)


# ------------------------------------------------------------
# 2. SALES DATA
# ------------------------------------------------------------

sales = {
    "Total Leads": 128,
    "Offers Submitted": 68,
    "Orders Received": 22,
    "Lead Conversion Rate (%)": 17,
    "Total Offers Value": 417.53,
    "Total Orders Value": 55.10,
    "Pending Billing": 84.01,
    "Billed Value": 216.12,
    "FY 2026-27 Sales": 300.10,
    "FY 2025-26 Sales": 606.26
}

sales_data = pd.Series(sales, name="Value")


# ------------------------------------------------------------
# 3. INVENTORY DATA
# ------------------------------------------------------------

inventory = {
    "Raw Material": 15.81,
    "Semifinished Products": 110.10,
    "Finished Products": 1.68,
    "Dead Inventory": 9.70,
    "Tools & Consumables": 0.85
}

inventory_data = pd.Series(inventory, name="Value")


# ------------------------------------------------------------
# 4. ACCOUNTING DATA
# ------------------------------------------------------------

accounting = {
    "Debtors": 96.28,
    "Creditors": 22.80,
    "Bank Balance Available": 147.50,
    "Bank Balance Negative": 99.40
}

accounting_data = pd.Series(accounting, name="Value")


# ------------------------------------------------------------
# 5. SAVE GRAPH FUNCTION
# ------------------------------------------------------------

def save_graph(title):
    """
    Save the current Matplotlib graph as a PNG file.
    Every graph receives a unique filename.
    """
    global graph_counter

    graph_counter += 1

    safe_title = (
        title.lower()
        .replace(" ", "_")
        .replace("/", "_")
        .replace("-", "_")
    )

    filename = f"{graph_counter:02d}_{safe_title}.png"
    filepath = os.path.join(GRAPH_DIR, filename)

    plt.savefig(
        filepath,
        dpi=200,
        bbox_inches="tight"
    )

    print(f"Graph saved: {filename}")

    # Close the graph to avoid memory issues.
    # The graph image remains saved on disk.
    plt.close()


# ------------------------------------------------------------
# 6. CALCULATE BUSINESS KPIs
# ------------------------------------------------------------

total_leads = sales["Total Leads"]
total_offers = sales["Offers Submitted"]
total_orders = sales["Orders Received"]

lead_conversion = (total_orders / total_leads) * 100

offer_conversion = (
    (total_orders / total_offers) * 100
    if total_offers else 0
)

order_offer_value_ratio = (
    sales["Total Orders Value"]
    / sales["Total Offers Value"]
    * 100
)

total_inventory = inventory_data.sum()

debtor_creditor_difference = (
    accounting["Debtors"] - accounting["Creditors"]
)

fy_sales_difference = (
    sales["FY 2025-26 Sales"]
    - sales["FY 2026-27 Sales"]
)

fy_sales_change_percent = (
    fy_sales_difference
    / sales["FY 2025-26 Sales"]
    * 100
)

print("\nBUSINESS KPI SUMMARY")
print("-" * 60)

print(f"Total Leads: {total_leads}")
print(f"Offers Submitted: {total_offers}")
print(f"Orders Received: {total_orders}")
print(f"Calculated Lead Conversion: {lead_conversion:.2f}%")
print(f"Offer-to-Order Conversion: {offer_conversion:.2f}%")
print(f"Order Value / Offer Value: {order_offer_value_ratio:.2f}%")
print(f"Calculated Inventory Total: {total_inventory:.2f} Lakhs")
print(
    f"Debtors - Creditors: "
    f"{debtor_creditor_difference:.2f} Lakhs"
)
print(
    f"Previous FY sales minus current FY sales: "
    f"{fy_sales_difference:.2f} Lakhs"
)
print(
    f"Difference as % of previous FY sales: "
    f"{fy_sales_change_percent:.2f}%"
)


# ------------------------------------------------------------
# 7. GRAPH 1: SALES COUNTS BAR CHART
# ------------------------------------------------------------

sales_counts = pd.Series({
    "Leads": total_leads,
    "Offers Submitted": total_offers,
    "Orders Received": total_orders
})

plt.figure(figsize=(9, 6))

ax = sns.barplot(
    x=sales_counts.index,
    y=sales_counts.values
)

ax.bar_label(ax.containers[0], padding=3)

plt.title("October Sales Funnel - Counts")
plt.xlabel("Sales Stage")
plt.ylabel("Number of Leads / Offers / Orders")
plt.tight_layout()

save_graph("sales_counts_bar_chart")


# ------------------------------------------------------------
# 8. GRAPH 2: SALES VALUE BAR CHART
# ------------------------------------------------------------

sales_values = pd.Series({
    "Offers Value": sales["Total Offers Value"],
    "Orders Value": sales["Total Orders Value"],
    "Pending Billing": sales["Pending Billing"],
    "Billed Value": sales["Billed Value"]
})

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    x=sales_values.index,
    y=sales_values.values
)

ax.bar_label(ax.containers[0], fmt="%.2f", padding=3)

plt.title("Sales and Billing Values")
plt.xlabel("Category")
plt.ylabel("Value (Lakhs)")
plt.xticks(rotation=15)
plt.tight_layout()

save_graph("sales_billing_values")


# ------------------------------------------------------------
# 9. GRAPH 3: HORIZONTAL SALES FUNNEL
# ------------------------------------------------------------

funnel_data = pd.DataFrame({
    "Stage": [
        "Leads Generated",
        "Offers Submitted",
        "Orders Received"
    ],
    "Count": [
        total_leads,
        total_offers,
        total_orders
    ]
})

plt.figure(figsize=(10, 5))

ax = sns.barplot(
    data=funnel_data,
    x="Count",
    y="Stage"
)

ax.bar_label(ax.containers[0], padding=4)

plt.title("Lead-to-Order Funnel")
plt.xlabel("Count")
plt.ylabel("Sales Stage")
plt.tight_layout()

save_graph("sales_funnel")


# ------------------------------------------------------------
# 10. GRAPH 4: INVENTORY BAR CHART
# ------------------------------------------------------------

plt.figure(figsize=(11, 6))

ax = sns.barplot(
    x=inventory_data.values,
    y=inventory_data.index
)

ax.bar_label(ax.containers[0], fmt="%.2f", padding=4)

plt.title("October Inventory Distribution")
plt.xlabel("Value (Lakhs)")
plt.ylabel("Inventory Category")
plt.tight_layout()

save_graph("inventory_bar_chart")


# ------------------------------------------------------------
# 11. GRAPH 5: INVENTORY PIE CHART
# ------------------------------------------------------------

plt.figure(figsize=(9, 9))

plt.pie(
    inventory_data.values,
    labels=inventory_data.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Inventory Composition")
plt.tight_layout()

save_graph("inventory_pie_chart")


# ------------------------------------------------------------
# 12. GRAPH 6: ACCOUNTING BAR CHART
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    x=accounting_data.index,
    y=accounting_data.values
)

ax.bar_label(ax.containers[0], fmt="%.2f", padding=3)

plt.title("October Accounting Overview")
plt.xlabel("Accounting Metric")
plt.ylabel("Value (Lakhs)")
plt.xticks(rotation=15)
plt.tight_layout()

save_graph("accounting_bar_chart")


# ------------------------------------------------------------
# 13. GRAPH 7: DEBTORS VS CREDITORS PIE CHART
# ------------------------------------------------------------

debt_credit_data = pd.Series({
    "Debtors": accounting["Debtors"],
    "Creditors": accounting["Creditors"]
})

plt.figure(figsize=(8, 8))

plt.pie(
    debt_credit_data.values,
    labels=debt_credit_data.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Debtors vs Creditors")
plt.tight_layout()

save_graph("debtors_creditors_pie_chart")


# ------------------------------------------------------------
# 14. GRAPH 8: FINANCIAL POSITION
# ------------------------------------------------------------

financial_data = pd.Series({
    "Debtors": accounting["Debtors"],
    "Creditors": accounting["Creditors"],
    "Available Bank Balance": accounting["Bank Balance Available"],
    "Negative Bank Balance": accounting["Bank Balance Negative"]
})

plt.figure(figsize=(11, 6))

ax = sns.barplot(
    x=financial_data.index,
    y=financial_data.values
)

ax.bar_label(ax.containers[0], fmt="%.2f", padding=3)

plt.title("Financial Position Overview")
plt.xlabel("Financial Metric")
plt.ylabel("Value (Lakhs)")
plt.xticks(rotation=15)
plt.tight_layout()

save_graph("financial_position")


# ------------------------------------------------------------
# 15. GRAPH 9: HISTOGRAM
# ------------------------------------------------------------

# This histogram explores the distribution of the supplied
# numeric metrics. These metrics are from different business
# categories, so the histogram is descriptive only.

numeric_values = pd.concat([
    sales_data,
    inventory_data,
    accounting_data
])

plt.figure(figsize=(10, 6))

sns.histplot(
    data=numeric_values,
    bins=10,
    kde=False
)

plt.title("Distribution of Numeric Business Values")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.tight_layout()

save_graph("numeric_values_histogram")


# ------------------------------------------------------------
# 16. GRAPH 10: CONVERSION RATE COMPARISON
# ------------------------------------------------------------

conversion_data = pd.Series({
    "Leads to Orders": lead_conversion,
    "Offers to Orders": offer_conversion,
    "Order Value / Offer Value": order_offer_value_ratio
})

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    x=conversion_data.index,
    y=conversion_data.values
)

ax.bar_label(ax.containers[0], fmt="%.2f%%", padding=3)

plt.title("Sales Conversion Metrics")
plt.xlabel("Metric")
plt.ylabel("Percentage (%)")
plt.xticks(rotation=15)
plt.tight_layout()

save_graph("conversion_rates")


# ------------------------------------------------------------
# 17. GRAPH 11: FINANCIAL KPI COMPARISON
# ------------------------------------------------------------

financial_kpis = pd.Series({
    "Inventory Total": total_inventory,
    "Debtors - Creditors": debtor_creditor_difference,
    "Available Bank Balance": accounting["Bank Balance Available"],
    "Negative Bank Balance": accounting["Bank Balance Negative"]
})

plt.figure(figsize=(11, 6))

ax = sns.barplot(
    x=financial_kpis.index,
    y=financial_kpis.values
)

ax.bar_label(ax.containers[0], fmt="%.2f", padding=3)

plt.title("Selected Business and Financial KPIs")
plt.xlabel("KPI")
plt.ylabel("Value (Lakhs)")
plt.xticks(rotation=15)
plt.tight_layout()

save_graph("business_financial_kpis")


# ------------------------------------------------------------
# 18. GRAPH 12: FINANCIAL YEAR SALES COMPARISON
# ------------------------------------------------------------

fy_data = pd.Series({
    "FY 2025-26": sales["FY 2025-26 Sales"],
    "FY 2026-27": sales["FY 2026-27 Sales"]
})

plt.figure(figsize=(8, 6))

ax = sns.barplot(
    x=fy_data.index,
    y=fy_data.values
)

ax.bar_label(ax.containers[0], fmt="%.2f", padding=3)

plt.title("Financial Year Sales Comparison")
plt.xlabel("Financial Year")
plt.ylabel("Sales (Lakhs)")
plt.tight_layout()

save_graph("financial_year_sales_comparison")


# ------------------------------------------------------------
# 19. SAVE DATA TO CSV
# ------------------------------------------------------------

sales_data.to_csv(
    os.path.join(OUTPUT_DIR, "sales_data.csv"),
    header=True
)

inventory_data.to_csv(
    os.path.join(OUTPUT_DIR, "inventory_data.csv"),
    header=True
)

accounting_data.to_csv(
    os.path.join(OUTPUT_DIR, "accounting_data.csv"),
    header=True
)

# Combine the supplied metrics into one CSV.
# Each row is a metric; values are not all comparable.

all_metrics = pd.concat([
    sales_data.rename("Sales"),
    inventory_data.rename("Inventory"),
    accounting_data.rename("Accounting")
], axis=0)

all_metrics.to_csv(
    os.path.join(OUTPUT_DIR, "business_data.csv"),
    header=["Value"]
)


# ------------------------------------------------------------
# 20. SAVE BUSINESS KPI SUMMARY
# ------------------------------------------------------------

kpi_summary = pd.Series({
    "Calculated Lead Conversion (%)": lead_conversion,
    "Offer-to-Order Conversion (%)": offer_conversion,
    "Order Value / Offer Value (%)": order_offer_value_ratio,
    "Calculated Inventory Total (Lakhs)": total_inventory,
    "Debtors minus Creditors (Lakhs)": debtor_creditor_difference,
    "FY Sales Difference (Lakhs)": fy_sales_difference,
    "FY Difference as % of Previous FY Sales": fy_sales_change_percent
}, name="Value")

kpi_summary.to_csv(
    os.path.join(OUTPUT_DIR, "business_kpi_summary.csv"),
    header=True
)


# ------------------------------------------------------------
# 21. CREATE ZIP FILE
# ------------------------------------------------------------

zip_base = os.path.join(BASE_DIR, "business_graphs")

zip_path = shutil.make_archive(
    zip_base,
    "zip",
    root_dir=OUTPUT_DIR
)


# ------------------------------------------------------------
# 22. FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("ANALYSIS COMPLETED SUCCESSFULLY!")
print("=" * 60)

print(f"Total graphs generated: {graph_counter}")
print(f"Graphs folder: {GRAPH_DIR}")
print(f"CSV files folder: {OUTPUT_DIR}")
print(f"ZIP file: {zip_path}")

print("\nShare business_graphs.zip to send all graphs.")
print("Open the outputs folder to access individual PNG files.")
```

```python
import os
import glob
from PIL import Image
from reportlab.lib.pagesizes import A4
from reportlab.platypus import (
    SimpleDocTemplate, Image as PDFImage,
    Paragraph, Spacer, PageBreak
)
from reportlab.lib.styles import getSampleStyleSheet
from reportlab.lib.units import inch

# Folder containing your saved graphs
graph_folder = os.path.abspath("outputs/graphs")

# Find all graph images
graph_files = sorted(
    glob.glob(os.path.join(graph_folder, "*.png"))
)

if not graph_files:
    print("No graphs found. Check your outputs/graphs folder.")
else:
    pdf_path = os.path.abspath("Business_Graphs_Report.pdf")

    document = SimpleDocTemplate(
        pdf_path,
        pagesize=A4,
        rightMargin=40,
        leftMargin=40,
        topMargin=40,
        bottomMargin=40
    )

    styles = getSampleStyleSheet()
    story = []

    for index, filepath in enumerate(graph_files):
        title = os.path.basename(filepath).replace("_", " ")

        story.append(Paragraph(title, styles["Heading2"]))
        story.append(Spacer(1, 15))

        with Image.open(filepath) as img:
            width, height = img.size

        max_width = 7 * inch
        max_height = 8 * inch

        scale = min(
            max_width / width,
            max_height / height
        )

        story.append(
            PDFImage(
                filepath,
                width=width * scale,
                height=height * scale
            )
        )

        if index < len(graph_files) - 1:
            story.append(PageBreak())

    document.build(story)

    print("PDF created successfully!")
    print("Total graphs:", len(graph_files))
    print("File location:", pdf_path)
```

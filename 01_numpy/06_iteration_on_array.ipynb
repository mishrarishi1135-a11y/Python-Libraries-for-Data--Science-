```python
# Iteration on numpy array  (loops in array)
import numpy as np
a1 = np.arange(11)
a2 = np.arange(12).reshape(3,4)
a3 = np.arange(27).reshape(3,3,3)
a3
```

```python
a1
for i in a1:
    print(i)
```

```python
# In 2D, print a one row at one time
for i in a2:
    print(i)
```

```python
# In 3D, print a 2D array at one time
for i in a3:
    print(i)
```

```python
# how many dimension array, we use this (nditer) for accessing the each and every element
for i in np.nditer(a3):
    print(i)      # 3d --> 2D --> 1D
```

```python
# array functions --> you don't need to make any function for mathematical calculation.
# also read the documentation of numpy...
import numpy as np
a1 = np.random.random((3,3))
a1 = np.round(a1*100)
a1
```

```python
# max/min/sum/prod
# 0 --> column and 1 --> row
np.max(a1)
```

```python
np.min(a1)
```

```python
np.sum(a1)
```

```python
np.prod(a1)
```

```python
np.max(a1,axis =0)
```

```python
# mean/median/std/var
np.mean(a1)
```

```python
np.median(a1,axis=1)
```

```python
np.var(a1)
```

```python
# trigonometric function --> we don't use this function but see only for knowledge not use in data science.
np.sin(a1)
```

```python
# dot product: condition --> column no of 1st matrix = row no of 2nd matrix
a2 = np.arange(12).reshape(3,4)
a3 = np.arange(12,24).reshape(4,3)
np.dot(a2,a3)
```

```python
# log and exponents
np.log(a1)
```

```python
np.exp(a2)
```

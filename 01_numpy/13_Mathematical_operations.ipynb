```python
# Working with mathematical formulas
# we know python has much more built-in function.
# now we can solve any mathematical problem by using these functions.
# sigmoid --> much more use in logistic regression and also in deep learning
#   S(x) = 1 / (1 + e^-x); formula of sigmoid
import numpy as np
def sigmoid(array):
    return 1/ (1 + np.exp(-(array)))
a = np.arange(10)
sigmoid(a)
# sigmoid lies between 0 and 1.
```

```python
# now we calculate mean squared error on both array
actual = np.random.randint(1,50,25)
predicted = np.random.randint(1,50,25)

def mse(actual, predicted):
    return np.mean((actual - predicted)**2)

mse(actual,predicted)    # It will give the combined avg error.
```

```python
# binary cross entropy : for logistic regression
# it's a loss function which is used in binary classification problems where output is only 0 or 1.
# Generally :  lower loss = better prediction
import numpy as np
def binary_cross_entropy(actual,predicted):
    return -np.mean(actual * np.log(predicted) + (1 - actual) * np.log(1 - predicted))
actual = np.array([1,0,1,1])
predicted = np.array([0.9,0.2,0.8,0.6])

print(binary_cross_entropy(actual,predicted))

# BCE measure the quality of classification probability.
```

```python
# horizontal stacking: add two or more numpy arrays in horizontal
import numpy as np
a1 = np.arange(12).reshape(3,4)
a2 = np.arange(12,24).reshape(3,4)
a2
```

```python
np.hstack((a1,a2))
```

```python
np.hstack((a1,a2,a1,a2))
```

```python
# vertical stacking
np.vstack((a1,a2))
```

```python
np.vstack((a1,a2,a1,a2))
```

```python
# multiple data sources ke same data ko ek sath rkhne ke liye iska use krte hai
# some data from database and some data from api and also some data from webscraping(all data similar)
# Remember one point "shape must be same "

# np.stack() --> creates a new axis, unlike vstack and hstack, stack() adds a new dimension.
a = np.array([1,2,3])
b = np.array([4,5,6])
np.stack((a,b))
```

```python
# np.concatenate() --> joins along an existing axis
np.concatenate((a,b))
```

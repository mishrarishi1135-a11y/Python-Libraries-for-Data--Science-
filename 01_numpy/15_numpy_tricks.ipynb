```python
# numpy tricks(hidden power of numpy)
# np.sort() --> return a sorted copy of an array
import numpy as np
a = np.random.randint(1,100,15)
a
```

```python
b = np.random.randint(1,100,24).reshape(6,4)
b
```

```python
np.sort(b,axis = 0)   # column wise sorting
```

```python
# for sorting in descending order
np.sort(a)[::-1]
```

```python
# np.concatenate() function concatenate a sequence of arrays along an existing axis.
c = np.arange(6).reshape(2,3)
d = np.arange(6,12).reshape(2,3)
print(c)
print(d)
```

```python
np.concatenate((c,d),axis = 0)   # row wise concatenate
```

```python
np.concatenate((c,d),axis = 1)  # column wise concatenate
```

```python
# np.where() --> return the indices of elements in an input array where the given condition is satisfied.
a = np.random.randint(1,100,25)
a
```

```python
# find all indices with value greater than 50
np.where(a > 50)
# it gives the index value.
```

```python
# replace all values which is greater than 50 with zero
np.where(a > 50, 0,a)      # (a > 50, true then print 0 otherwise print a)
```

```python
# replace all even values with zero
np.where(a % 2 == 0,0,a)
```

```python
# np.corrcoef() --> return pearson product moment correlation co-efficient.
# correlation is show only in between -1 to 1.
salary = np.array([20000,40000,25000,35000,60000])
experience = np.array([1,3,2,4,2])
np.corrcoef(salary,experience)
```

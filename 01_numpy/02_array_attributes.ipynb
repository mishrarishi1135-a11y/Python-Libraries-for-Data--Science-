```python
# first we create a three array for all this section
import numpy as np
a1 = np.arange(10)    # vector array
a2 = np.arange(12,dtype=float).reshape(3,4)    # Matrix array
a3 = np.arange(8).reshape(2,2,2)      # tensor array
a3
```

```python
# ndim --> give the dimension of given array
a3.ndim
```

```python
# shape --> each dimension have how many items?
a1.shape
```

```python
print(a3.shape)
a3
```

```python
# size --> total no of items in given array
a3.size
```

```python
# item size --> how much space occupy by every item in memory
a1.itemsize     # int32 contain 4 bytes and int64 contain 8 bytes.
```

```python
# dtype --> what is data types of item.
print(a1.dtype)
print(a2.dtype)
print(a3.dtype)
```

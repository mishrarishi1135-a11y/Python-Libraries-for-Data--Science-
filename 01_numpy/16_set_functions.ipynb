```python
# set functions
import numpy as np
m = np.array([1,2,3,4,5])
n = np.array([3,4,5,6,7])
np.union1d(m,n)
```

```python
np.intersect1d(m,n)
```

```python
np.setdiff1d(m,n)
```

```python
np.setdiff1d(n,m)
```

```python
np.setxor1d(m,n)   # remove common values which are present in both
```

```python
# np.in1d(m,1)   # like a membership operator
# this function is remove from numpy 2.0
# so we use np.in1d() in place of np.isin()
import numpy as np
m = np.array([1,2,3,4,5])
n = np.array([2,4,6])

print(np.isin(m,n))
```

```python
# np.setxor1d() --> remove common values which are present in both.
np.setxor1d(m,n)
```

```python
# np.in1d() --> for search any value in a given array in the boolean form
#  np.in1d()  # this was remove from the numpy 2.0
# so we use np.isin()

a = np.array([1,2,3,4,5,6])
b = np.array([2,4,6])

print(np.isin(a, b))
```

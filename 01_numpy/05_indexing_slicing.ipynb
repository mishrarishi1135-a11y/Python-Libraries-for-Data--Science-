```python
# indexing and slicing
import numpy as np
a1 = np.arange(10)   # 1D array
a2 = np.arange(12).reshape(3,4)   # 2D array
a3 = np.arange(8).reshape(2,2,2)   # 3D array
a1
```

```python
a1[0]
```

```python
a2
```

```python
a2[1,2]   #[row,column]
```

```python
a3[1,0,1]    # [in which 2D array - 1,row,column]
```

```python
a3[0,0,0]
```

```python
# slicing --> access multiple items at a one time
a1[2:5]    #[include:excluded]
```

```python
a2
```

```python
a2[0,:]   # produce column
```

```python
a2[2,:]   # produce column
```

```python
a2[:,2]   # produce row
```

```python
a2[1:,1:3]
```

```python
a2[::2,::3]
```

```python
a2[::2,1::2]
```

```python
a2[1,::3]
```

```python
a2[0:2,1::]     #[kon sa rows,kon sa column] just ask with you.
```

```python
a3 = np.arange(27).reshape(3,3,3)   # This numpy array consist of three 2-D array.
a3
```

```python
a3[1]
```

```python
a3[::2]      # :: --> sabko nikal lo aur alternate ki digit likh do.
```

```python
a3[0,1,:]        # we use (:) for getting all the value of that particular column.
```

```python
a3[1,:,1]        # [2nd 2D matrix, all rows,mid column]
```

```python
a3[2,1:,1:]        #[3rd 2D matrix, 2nd row se end tk,2nd column se end tak]
```

```python
a3[::2,0,::2]               # [alternate numpy arrays,want only 1st row,want only 1st column and 3rd column]
```

```python
# normal indexing and slicing
import numpy as np
a = np.arange(12).reshape(3,4)
a
```

```python
# if we access any element so just use rom and column number
a[1,2]
```

```python
# fancy indexing for fetch items -> accessing multiple specific position elements at a one time
import numpy as np
a = np.array([10, 20 ,30, 40, 50])
print(a[[0,2,4]])
```

```python
# 2-D array
a2 = np.arange(12).reshape(3,4)
a2
```

```python
# if we want 5, 2, 11
print(a2[[1,0,2],[1,2,3]])
```

```python
# why this?
# Row index         column index    element
#    1                  1             5
#    0                  2             2
#    2                  3             11
# in fancy indexing we just provide a list of index of elements.
```

```python
# Boolean indexing
a = np.random.randint(1,100,24).reshape(6,4)
a
```

```python
# find all numbers greater than 50 in array a
a > 50
```

```python
# if we want only those numbers which are greater tha 50
a[a > 50]
```

```python
# find out even numbers
a % 2 == 0
```

```python
a[a % 2 == 0]
```

```python
# find all numbers greater than 50 and are even
(a > 50) & (a % 2 == 0)
```

```python
a[(a > 50) & (a % 2 == 0)]
```

```python
# find all numbers that are not divisible by 7
import numpy as np
a1 = np.random.randint(100,200,24).reshape(6,4)
a1
```

```python
a1[a1 % 7 != 0]
```

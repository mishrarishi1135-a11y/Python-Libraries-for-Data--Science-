```python
# splitting is just opposite of the stacking
import numpy as np
a1 = np.arange(12).reshape(3,4)
a1
```

```python
# horizontal splitting
np.hsplit(a1,2)       # here 2 is how many parts we divide this array
```

```python
np.vsplit(a1,3)    # if we write 2 then it show error because it cut only in equal parts.
```

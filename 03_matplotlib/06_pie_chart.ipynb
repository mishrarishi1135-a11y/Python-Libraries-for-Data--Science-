```python
import matplotlib.pyplot as plt
import pandas as pd
df = pd.read_csv('../datasets/gayle-175.csv')
plt.pie(df['batsman_runs'],labels=df['batsman'])
plt.show()
```

```python
# for percentage we sue a parameter --> autopct
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%')
plt.show()
# you also add color according to your specification by using 'color' parameter.
```

```python
# explode shadow --> for showing specially
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',explode=[0.1,0,0,0,0,0.1],shadow=True)
plt.show()
# 0.1 --> jis ko bahar lana hai
# 0 --> jis ko bahar nhi lana hai
```

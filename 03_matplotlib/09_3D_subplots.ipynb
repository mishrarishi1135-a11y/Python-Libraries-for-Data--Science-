```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
batter = pd.read_csv('../datasets/batter.csv')
batter
```

```python
# now we can plot a 3D graph(avg,runs and strike rate)
fig = plt.figure()

ax = plt.subplot(projection='3d')   # By using projection we can create a 3 dimension graph.
```

```python
fig = plt.figure()
ax = plt.subplot(projection='3d')

ax.scatter3D(batter['runs'],batter['avg'],batter['strike_rate'])
```

```python
fig = plt.figure()
ax = plt.subplot(projection='3d')

ax.scatter3D(batter['runs'],batter['avg'],batter['strike_rate'])

# You can also give a title and axis label.
ax.set_title('IPL batsman analysis')

ax.set_xlabel('runs')
ax.set_ylabel('avg')
ax.set_zlabel('strike_rate')
```

```python
# 3D line plot

x = [0,1,5]
y = [0,10,13]
z = [0,13,20]
fig = plt.figure()
ax = plt.subplot(projection='3d')

ax.scatter3D(x,y,z)   # This gives only points not adding these points.
# This is a scatter plot but adding these points we can use plot3D.
ax.plot3D(x,y,z,color='red')
```

```python
# Surface plots in machine learning algorithms
# we plot this graph --> f(x) = x^2 + y^2
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)    # linspace will generate points.
```

```python
xx, yy = np.meshgrid(x,y)
xx.shape
```

```python
z = xx**2 + yy**2
z.shape
```

```python
fig = plt.figure(figsize=(12,8))
ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
z = np.sin(xx) + np.cos(yy)
fig = plt.figure(figsize=(12,8))
ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

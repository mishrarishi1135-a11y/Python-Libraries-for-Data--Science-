```python
# can be used on a dataframe as well
import seaborn as sns
tips = sns.load_dataset('tips')
tips
```

```python
tips['size'] = tips['size'] * 100
```

```python
tips.head()
```

```python
# Scatter plot -> labels -> markers -> figsize -> color -> cmap
tips.plot(kind='scatter',x='total_bill',y='tip',title='cost analysis',color='red',marker='+',figsize=(10,6),s = 'size')
```

```python
# If you want a data distribution in gender so use 'c' parameter but now you remove the color parameter because it creates confusion.
tips.plot(kind='scatter',x='total_bill',y='tip',title='cost analysis',c='sex',marker='+',figsize=(10,6),s = 'size')
```

```python
# For better understanding use 'cmap' parameter
tips.plot(kind='scatter',x='total_bill',y='tip',title='cost analysis',c='sex',cmap='viridis',marker='+',figsize=(10,6),s = 'size')
```

```python
# pie chart -> single and multiple
df = pd.DataFrame(
    {
        'batsman':['Dhavan','Rohit','Kohli','Sky','Pandya',"pant"],
        "match1":[120,90,35,45,12,10],
        'match2':[0,1,123,130,34,45],
        'match3':[50,24,145,45,10,90]
    }
)

df.head()
```

```python
df['match1'].plot(kind='pie',labels=df['batsman'].values,autopct='%0.1f%%')
```

```python
# multiple pie charts
df[['match1','match2','match3']].plot(kind='pie',subplots=True,figsize=(15,8))
```

```python
# On multi index dataframe
# using tips
tips
```

```python
tips.head()
```

```python
tips.pivot_table(index='day',columns='sex',values='total_bill',aggfunc='mean')
```

```python
tips.pivot_table(index='day',columns='sex',values='total_bill',aggfunc='mean').plot(kind="line")
```

```python
# now we want a time along with day so we can pass a list
tips.pivot_table(index=['day','time'],columns='sex',values='total_bill',aggfunc='mean').plot(kind="line")
```

```python
# Now we create a categories in columns
tips.pivot_table(index=['day','time'],columns=['sex','smoker'],values='total_bill',aggfunc='mean').plot(kind="line")
```

```python
# You can also plot a bar chart for this
tips.pivot_table(index=['day','time'],columns=['sex','smoker'],values='total_bill',aggfunc='mean').plot(kind="bar")
```

```python
# for creating pie chart for this you also use subplots=True
tips.pivot_table(index=['day','time'],columns=['sex','smoker'],values='total_bill',aggfunc='mean').plot(kind="pie",subplots=True,figsize=(20,10))
```

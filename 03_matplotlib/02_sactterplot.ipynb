```python
# plt.scatter simple function
import numpy as np
x = np.linspace(-10,10,50)
x
y = 10*3 + np.random.randint(0,300,50)
y
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
plt.scatter(x,y)
```

```python
# plt.scatter on pandas data
df = pd.read_csv('../datasets/batter.csv')
df
```

```python
# now we plot a scatter plot of top 50 players where x=avg and y=strike rate
df = pd.read_csv('../datasets/batter.csv')
df = df.head(50)
df
```

```python
plt.scatter(df['avg'],df['strike_rate'])
```

```python
plt.scatter(df['avg'],df['strike_rate'])
plt.title('avg and SR analysis of top 50 batsman')
plt.xlabel('Average')
plt.ylabel('strike_rate')
```

```python
# also you can change the color and marker size for the graph
plt.scatter(df['avg'],df['strike_rate'],color='red',marker='+')
plt.title('avg and SR analysis of top 50 batsman')
plt.xlabel('Average')
plt.ylabel('strike_rate')
```

```python
# you can also control the size parameter.
tips = sns.load_dataset('tips')
tips
# here we have two numerical column and now we plot the scatter plot using these two numerical column.
```

```python
tips = sns.load_dataset('tips')
plt.scatter(tips['total_bill'],tips['tip'])
```

```python
# we can also control the size of dot like, more customer -> then larger dot and less customer -> smaller dot
tips = sns.load_dataset('tips')
plt.scatter(tips['total_bill'],tips['tip'],s=tips['size'])
```

```python
# for better visibility we can multiply the graph using any constant value
plt.scatter(tips['total_bill'],tips['tip'],s=tips['size']*20)
```

```python
# scatter plot using plt.plot() without using scatter function
plt.plot(tips['total_bill'],tips['tip'])  # 2D plot
```

```python
# But if you give a additional parameter then it show better
plt.plot(tips['total_bill'],tips['tip'],'o')
```

```python
# plt.plot() --> faster technique
# scatter.plot() --> slower technique
```

```python
# Bar chart -> x-axis: categorical column and y-axis:perform aggregate operation
# Use for Bi variate analysis
# Numerical vs categorical
# use case -> aggregate analysis of groups
```

```python
# simple bar chart
import matplotlib.pyplot as plt
children = [10,20,40,10,30]
colors = ['red','blue','green','yellow','pink']
plt.bar(colors,children)    # Horizontal bar chart
```

```python
# Two types of bar chart
import matplotlib.pyplot as plt
children = [10,20,40,10,30]
colors = ['red','blue','green','yellow','pink']
plt.barh(colors,children,color='black')
```

```python
# Color and label
import numpy as np
import pandas as pd
import seaborn as sns
df = pd.read_csv('../datasets/batsman_season_record.csv')
df
```

```python
plt.bar(df['batsman'],df['2015'])
```

```python
plt.bar(df['batsman'],df['2015'])
plt.bar(df['batsman'],df['2016'])
plt.bar(df['batsman'],df['2017'])   # It's not better because all the bar charts overlap.
# so we print the bar chart side by side for a particular year.
```

```python
# You can also control the width of bars.
plt.bar(df['batsman'],df['2015'],width=0.5)
```

```python
np.arange(df.shape[0])
```

```python
plt.bar(np.arange(df.shape[0]) - 0.5,df['2015'],width=0.75)
```

```python
plt.bar(np.arange(df.shape[0]) - 0.2,df['2015'],width=0.2)
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2)
plt.bar(np.arange(df.shape[0]) + 0.2,df['2017'],width=0.2)

plt.xticks(np.arange(df.shape[0]),df['batsman'])
plt.show()
```

```python
# If you have large name in x-axis then it would be overlap,when you have much more number of categories.
children = [10,20,40,10,30]
colors = ['red red red red','blue blue blue blue','green green green green','yellow yellow yellow yellow','pink pink pink pink']
plt.barh(colors,children,color='black')
plt.xticks(rotation='vertical')
plt.show()
```

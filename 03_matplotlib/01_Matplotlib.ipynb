```python
# matplotlib -> mother of all data visualization library
# Types of data --> Numerical data(age,temp,height) and categorical data(where we produce categories in data like gender,mobile brand in a group)
# Univariate analysis -> when we plot the graph on one column.
# Bi variate analysis -> when we plot the graph on two column.
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 2D line plot -> use for Bi-variate analysis
# Use case -> Time series data
# when we use this plot --> categorical > numerical and categorical > numerical

# Plotting a simple function
price = [48000,54000,57000,49000,47000,45000]     # numerical data
year = [2015,2016,2017,2018,2019,2020]            # categorical data
plt.plot(year,price)       #(x-axis,y-axis)
```

```python
# From a pandas dataframe
batsman = pd.read_csv('../datasets/sharma-kohli.csv')
batsman
```

```python
plt.plot(batsman['index'],batsman['V Kohli'])
```

```python
# Plotting multiple plots --> there is no upper limit for plotting the graph.
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
```

```python
# If you want to show the title of graph
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])

plt.title("Rohit sharma VS Virat Kohli career comparison")
plt.xlabel("season")
plt.ylabel("runs scored")
```

```python
# You also change the colors --> colors(hex) and line(width and style) and marker(size)
plt.plot(batsman['index'],batsman['V Kohli'],color='green')
plt.plot(batsman['index'],batsman['RG Sharma'],color='black')

plt.title("Rohit sharma VS Virat Kohli career comparison")
plt.xlabel("season")
plt.ylabel("runs scored")
# You also gives the html hex code for changing the colors of graph.
```

```python
# You have also a line style parameter
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='dashed')   # also use dotted, solid for line style.
plt.plot(batsman['index'],batsman['RG Sharma'],color='black',linestyle='dashed')

plt.title("Rohit sharma VS Virat Kohli career comparison")
plt.xlabel("season")
plt.ylabel("runs scored")
```

```python
# you also use linewidth parameter
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='solid',linewidth=3)
plt.plot(batsman['index'],batsman['RG Sharma'],color='black',linestyle='dashed',linewidth=2)

plt.title("Rohit sharma VS Virat Kohli career comparison")
plt.xlabel("season")
plt.ylabel("runs scored")
```

```python
# You can manipulate the points using marker
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='solid',linewidth=3,marker='D')
plt.plot(batsman['index'],batsman['RG Sharma'],color='black',linestyle='dashed',linewidth=2,marker='+')

plt.title("Rohit sharma VS Virat Kohli career comparison")
plt.xlabel("season")
plt.ylabel("runs scored")
```

```python
# If you want to print the label then you use the legend function
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='solid',linewidth=3,marker='D',label="virat")
plt.plot(batsman['index'],batsman['RG Sharma'],color='black',linestyle='dashed',linewidth=2,marker='+',label="rohit")

plt.title("Rohit sharma VS Virat Kohli career comparison")
plt.xlabel("season")
plt.ylabel("runs scored")

plt.legend()    # legend have also 'loc' - right upper, best, upper centre.
```

```python
# grid --> for betetr readability
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='solid',linewidth=3)
plt.plot(batsman['index'],batsman['RG Sharma'],color='black',linestyle='dashed',linewidth=2)

plt.title("Rohit sharma VS Virat Kohli career comparison")
plt.xlabel("season")
plt.ylabel("runs scored")
plt.grid()
```

```python
# show --> It is main function, by this actual graph plotting is possible.
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='solid',linewidth=3)
plt.plot(batsman['index'],batsman['RG Sharma'],color='black',linestyle='dashed',linewidth=2)

plt.title("Rohit sharma VS Virat Kohli career comparison")
plt.xlabel("season")
plt.ylabel("runs scored")
plt.grid()
plt.show()
```

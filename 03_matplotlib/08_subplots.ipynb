```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# A different way to plot graphs
batter = pd.read_csv('../datasets/batter.csv')
batter
```

```python
batter.head(5)
```

```python
plt.scatter(batter['avg'],batter['strike_rate'])
```

```python
# 2nd method
fig, ax = plt.subplots(figsize=(15,6))    # figure object which is graph and axis object like x and y axis.
# But reason is that now we can plot multiple graphs at a one time.
ax.scatter(batter['avg'],batter['strike_rate'])
# for giving title we use another function
ax.set_title("batsman scored runs")
# for label x and y axis
ax.set_xlabel('avg')
ax.set_ylabel('strike_rate')
# For control the size of graph
# use figsize() inside the subplots()

# for no technical information here we write
fig.show()
```

```python
# Now we want the print  two scatter plot side by side.
plt.subplots(nrows=2,ncols=1)
# Here you give single figure object but you have two axis object,(single figure but you can plot two graphs)
```

```python
fig, ax = plt.subplots(nrows=2,ncols=1,sharex=True,figsize=(10,6))   # by using sharex we show x points only in lower graph not in upper graph.
ax[0].scatter(batter['avg'],batter['strike_rate'])
ax[1].scatter(batter['avg'],batter['runs'])

ax[0].set_title('avg vs strike rate')
ax[0].set_ylabel('Strike rate')

ax[1].set_title('avg vs runs')
ax[1].set_ylabel('runs')
ax[1].set_xlabel('avg')    # because both graph have same x axis name.
```

```python
# now we plot 4 graphs
fig, ax = plt.subplots(nrows=2,ncols=2,figsize=(10,10))   # by using sharex we show x points only in lower graph not in upper graph.
ax[0,0].scatter(batter['avg'],batter['strike_rate'])
ax[0,1].scatter(batter['avg'],batter['runs'])
ax[1,0].hist(batter['avg'])
ax[1,1].hist(batter['runs'])
```

```python
fig = plt.figure()
ax1 = fig.add_subplot(2,1,1)    # 2 rows 1 column 1st graph --> 211
ax1.scatter(batter['avg'],batter['strike_rate'],color='red')

# Now if you want to add 2nd graph
ax2 = fig.add_subplot(2,1,2)    # 2 rows 1 column 2nd graph --> 212
ax2.hist(batter['runs'])
```

```python
# This is the way so you can add how many graphs.
fig = plt.figure()
ax1 = fig.add_subplot(2,2,1)
ax1.scatter(batter['avg'],batter['strike_rate'],color='red')

# Now if you want to add 2nd graph
ax2 = fig.add_subplot(2,2,2)
ax2.hist(batter['runs'])

ax3 = fig.add_subplot(2,2,3)
ax3.hist(batter['avg'])
```

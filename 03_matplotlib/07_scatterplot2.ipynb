```python
# Now we want to color encode in data for each particular value.
iris = pd.read_csv('../datasets/iris.csv')
iris.sample(5)
```

```python
iris['Species'] = iris['Species'].replace({'Iris-setosa':0,'Iris-versicolor':1,'Iris-virginica':2})
iris.sample(5)
# Now our flowers names re-assign.
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'])
# You can also give label
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.show()
# you can also use 'cmap' parameter for different themes in graph, cmap- color maps
```

```python
# you can also add color bar which tells that 'the value use by the color'
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='jet')
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar()   # don't use plt.show() when you use colorbar().
```

```python
# now we use alpha parameter - control the opacity level
# if alpha = 0.5 that means the line will be 50% transparent.
# Higher alpha - more visible - less transparent
# lower alpha - less visible - more transparent
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='jet',alpha=0.5)
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar()
```

```python
# plot size --> how can you control the size of your graph.
plt.figure(figsize=(15,7))     # Figsize wants a tuple (width of graph,height of graph)
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='jet',alpha=0.5)
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar()
```

```python
# annotations
batter = pd.read_csv('../datasets/batter.csv')
batter
```

```python
batter.head(5)
```

```python
batter.shape   # It tells 605 batsman and 4 columns.
```

```python
sample_batter = batter.head(100).sample(25,random_state=5)
sample_batter
```

```python
# Now we plot the scatter plot
plt.scatter(sample_batter['avg'],sample_batter['strike_rate'])
plt.xlabel('avg')
plt.ylabel('strike_rate')
```

```python
# but we don't know that which dot tell the ifo about which player and this problem is solve by annotation.
x = [1,2,3,4]
y = [5,6,7,8]
plt.scatter(x,y)
```

```python
# Now we wants to give the labels or name to these points.
x = [1,2,3,4]
y = [5,6,7,8]
plt.scatter(x,y)
plt.text(1,5,'Point 1')
plt.text(2,6,'Point 2')
plt.text(3,7,'Point 3')
plt.text(4,8,'Point 4')
# you use a text parameter(x coordinate, y coordinate, name as string)
```

```python
plt.scatter(sample_batter['avg'],sample_batter['strike_rate'])
plt.xlabel('avg')
plt.ylabel('strike_rate')

# Here we run a loop for 25 batsman because we don't write it manually
for i in range(sample_batter.shape[0]):
    plt.text(sample_batter['avg'].values[i],sample_batter['strike_rate'].values[i],sample_batter['batter'].values[i])
```

```python
# now you can change the size of graph for better readability
plt.figure(figsize=(15,7))
plt.scatter(sample_batter['avg'],sample_batter['strike_rate'])
plt.xlabel('avg')
plt.ylabel('strike_rate')

# Here we run a loop for 25 batsman because we don't write it manually
for i in range(sample_batter.shape[0]):
    plt.text(sample_batter['avg'].values[i],sample_batter['strike_rate'].values[i],sample_batter['batter'].values[i])
```

```python
# Now we decide the size of dot according to runs scored by batsman.
plt.figure(figsize=(15,7))
plt.scatter(sample_batter['avg'],sample_batter['strike_rate'],s = sample_batter['runs'])
plt.xlabel('avg')
plt.ylabel('strike_rate')

# Here we run a loop for 25 batsman because we don't write it manually
for i in range(sample_batter.shape[0]):
    plt.text(sample_batter['avg'].values[i],sample_batter['strike_rate'].values[i],sample_batter['batter'].values[i])
```

```python
# How can you plot horizontal and vertical lines on your graph.
plt.figure(figsize=(15,7))
plt.scatter(sample_batter['avg'],sample_batter['strike_rate'],s = sample_batter['runs'])
plt.xlabel('avg')
plt.ylabel('strike_rate')

plt.axhline(130,color='red')   # Horizontal line --> axh
plt.axvline(30,color='red')    # vertical line --> vxh

# Here we run a loop for 25 batsman because we don't write it manually
for i in range(sample_batter.shape[0]):
    plt.text(sample_batter['avg'].values[i],sample_batter['strike_rate'].values[i],sample_batter['batter'].values[i])

# you can improve your analysis by creating this type of lines in your graph.
```

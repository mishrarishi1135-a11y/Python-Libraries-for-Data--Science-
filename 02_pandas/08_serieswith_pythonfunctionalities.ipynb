```python
# membership operator
movies = pd.read_csv("../datasets/bollywood.csv")
print(movies)
```

```python
'Awara Paagal Deewana' in movies.values
```

```python
# because 'in' operator works on index if you want to work on value then write df.values
'Alia Bhatt' in movies.values
```

```python
# So membership operator works but only on index values not only values.
# similar you also run a loop

for i in movies:
    print(i)
```

```python
# using itertuples()
for row in movies.itertuples():
    print(f"Movie: {row.movie} | Lead: {row.lead}")
```

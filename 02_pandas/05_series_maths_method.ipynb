```python
# series maths method
import pandas as pd
df = pd.read_csv("../datasets/subs.csv")
series = df.squeeze()
print(series)
```

```python
# count --> it does not count missing values while size count the missing values.
series.count()   # gives you total no of items are present in the series only which they are non-missing values.
```

```python
# sum -- sum of all items
series.sum()
```

```python
series.product()   # In any day the no of subs is zero so it gives zero
```

```python
series.mean()
```

```python
series.median()
```

```python
series.mode()    # particular vo value so sb se jyada times data me ho.
# This zero is a index number of that value.
```

```python
print(series.std())
```

```python
print(series.var())
```

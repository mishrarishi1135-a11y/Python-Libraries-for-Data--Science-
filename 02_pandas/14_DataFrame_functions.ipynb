```python
# Find the last match played by virat kholi in delhi
ipl.head()
```

```python
# first we create a new column -> 'all player'
ipl['all player'] = ipl['Team1Players'] + ipl['Team2Players']
ipl.head()
```

```python
# Now we check inside the all players column 'virat is present or not'.
def did_kohli_play(players_list):
    return 'V Kohli' in players_list
```

```python
ipl['all player'].apply(did_kohli_play)
```

```python
ipl['did_kohli_play'] = ipl['all player'].apply(did_kohli_play)
ipl
```

```python
# now we find that matches where virat play the match. and match is in delhi.
ipl['did_kohli_play'] = ipl['all player'].apply(did_kohli_play)
ipl[(ipl['City'] == 'Delhi') & (ipl['did_kohli_play'] == True)]   # already sorted.
```

```python
# Now we find that recent match inside delhi and virat is present.
ipl[(ipl['City'] == 'Delhi') & (ipl['did_kohli_play'] == True)].drop_duplicates(subset=['City','did_kohli_play'],keep = 'first')
# we apply drop_duplicates() on the two column city and did_kohli_play.
```

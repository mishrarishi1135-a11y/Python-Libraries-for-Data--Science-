```python
# here we learn about the "Data Frame" object in Pandas.
# Jis ko aap table bulate ho Pandas usko DataFrame bolta hai.
# It is a two dimensional data structure that stores data in rows and columns, just like an excel spreadsheet and database table.
# Also single column is known as 'series' object and each row of dataframe is also a series.
import numpy as np
import pandas as pd
```

```python
# Creating Data Frame
# 1. using lists ; here we create a 2D list for this.
student_data =[
    [100,80,10],
    [90,70,7],
    [120,100,14],
    [80,50,2]
]
pd.DataFrame(student_data,columns=['IQ','Marks','Package'])
```

```python
# same dataframe using dictionary
student_dict ={
    'IQ': [100,90,120,80,0,0],
    'Marks': [80,70,100,50,0,0],
    'Package': [10,7,14,2,0,0]
}
students = pd.DataFrame(student_dict)
students
```

```python
# Now Understand how can we fetch row from DataFrame
student_dict ={
    'Name': ['nitish','Anu','rock','hil','pie','neil'],
    'IQ': [100,90,120,80,0,0],
    'Marks': [80,70,100,50,0,0],
    'Package': [10,7,14,2,0,0]
}
students = pd.DataFrame(student_dict)
students
```

```python
# set_index() -> It create a index from column, if you want.
students.set_index('Name')
# Here we create a Name column as a index.
```

```python
# For permanent change
students.set_index("Name",inplace=True)
```

```python
students
```

```python
# Lets learn how can we fetch a row from DataFrame.
# here we learn two methods -> iloc(searches using index positions) and loc(searches using index labels)
```

```python
# single row
movies.iloc[0] # It actually a series
```

```python
movies.iloc[4]
```

```python
# Fetch multiple rows
movies.iloc[0:5]  # But now it's a DataFrame.
```

```python
movies.iloc[5:15:2]  # If you want alternate movies
```

```python
movies.iloc[::-1]   # reverse the all dataframe.
```

```python
# fancy indexing
movies.iloc[[0,4,5]] # when you have no any such pattern for fetching the rows.
```

```python
# but now we understand 'loc'
students.loc['nitish']
```

```python
students.loc['Anu']
```

```python
# For multiple items
students.loc['nitish':'neil']
```

```python
# In 'iloc' -> last is excluded but  in 'loc' -> last is included.
students.loc['nitish':'neil':2]   # for alternate print
```

```python
# Here you can also use fancy indexing
students.loc[['nitish','pie','neil']]
```

```python
# You can also use iloc.
students.iloc[0]
```

```python
students.iloc[0:5]  # You can also see 'iloc' exclude the last value.
```

```python
students.iloc[[0,4,5]]
```

```python
# DataFrame.iloc[rows,columns]
movies.iloc[0:5,0:3]    # Here last is excluded.
```

```python
# sort-values  --> sort all the values
df = pd.read_csv("../datasets/kohli_ipl.csv")
series = df["runs"]
series.sort_values()
```

```python
# if you want to sort in descending order
series.sort_values(ascending=False)
# sort_values temporary changes krta hai original me koi change nhi aata.
# if you to change in original then (inplace=True)
```

```python
series.sort_values(ascending=False).head(1).values[0]   # gives numpy array
```

```python
series.sort_values(inplace=True)  # original series object me changes kr de ta hai which is not desired.
```

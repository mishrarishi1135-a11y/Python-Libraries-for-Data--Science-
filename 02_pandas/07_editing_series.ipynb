```python
# using indexing
import numpy as np
import pandas as pd
marks = {
    'math': 98,
    'english':97,
    'Hindi':100,
    'Physics':97
}
marks_series = pd.Series(marks,name='Anuj ke Marks')    # In dictionary, key becomes index and value becomes value
marks_series
```

```python
marks_series["english"] = 100
marks_series
```

```python
# what if an index does not exists
marks_series['sst'] = 98
```

```python
# slicing
runs = ([13,24,56,78,100])
runs_series = pd.Series(runs)
runs_series
```

```python
runs_series[2:4] = [100,100]    # edit multiple items
runs_series
```

```python
# fancy indexing
runs_series[[0,3,4]] = [0,0,0]
runs_series
```

```python
# What is pandas ??
# Pandas is a fast powerful, flexible and easy to use open source data analysis and manipulation tool, built on top of the python programming language.
# Pandas Series: It is like a column in table. It is a 1D array holding data of any type.
# Importing pandas
import numpy as np
import pandas as pd
```

```python
# series from list
import pandas as pd
my_list = [10,20,30,40,50]
series = pd.Series(my_list)   # By default index will be 0,1,2,3,4
print(series)
```

```python
# string
country = ['India','USA','China','Uk','Japan']
pd.Series(country)      # you can see your first series object
# simple words me numpy treat strings as a object.
```

```python
# Integers
runs = [13,24,56,78,93]
pd.Series(runs)
```

```python
# till then index create automatic in series. but you also create you 'custom index'.
marks = [88,98,97,96]
subjects = ['Hindi','english','Mathematics','Physics']   # we want subjects as a index.
pd.Series(marks,index = subjects)
```

```python
# if you want then you give a name to your series object.
marks = pd.Series(marks,index=subjects,name='Anuj Marks')
marks
```

```python
# series from dict
marks = {
    'math': 98,
    'english':97,
    'Hindi':100,
    'Physics':97
}
marks_series = pd.Series(marks)    # In dictionary, key becomes index and value becomes value
marks_series
```

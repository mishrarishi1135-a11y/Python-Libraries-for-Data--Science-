```python
# Series attributes
import numpy as np
import pandas as pd
marks = {
    'math': 98,
    'english':97,
    'Hindi':100,
    'Physics':97
}
marks_series = pd.Series(marks,name='Anuj ke Marks')    # In dictionary, key becomes index and value becomes value
marks_series
```

```python
# size
marks_series.size
```

```python
# dtype --> tell the data type of items which are given in series
marks_series.dtype
```

```python
# name
marks_series.name
```

```python
# is_unique --> if item repeat then it show false.
marks_series.is_unique
```

```python
pd.Series([1,2,3,4,5]).is_unique
```

```python
# index --> produces all the index objects
marks_series.index
```

```python
# values --> produces all values inside a numpy array
marks_series.values
```

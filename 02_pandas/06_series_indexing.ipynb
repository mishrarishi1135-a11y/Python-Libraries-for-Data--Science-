```python
# integer indexing
import pandas as pd
df = pd.read_csv("../datasets/subs.csv")
series = df.squeeze()
print(series)
```

```python
# pandas create a internal index which is 0,1,2,3,4,5.. while it have a value type indexing.
df = pd.read_csv("../datasets/bollywood.csv",index_col="movie")
series = df.iloc[:, 0]
print(series)
```

```python
series['Uri: The Surgical Strike'] # Also we use series[0]
```

```python
# slicing -- fetch multiple items at a one time
series[5:16]   # last excluded
```

```python
# Negative slicing
series[-5:]
```

```python
# if you want alternate movies
series[::2]
```

```python
# fancy indexing
series[[1,3,4,5]]   # we fetch values according to our need.
```

```python
# indexing with labels --> fancy indexing
series["22 Yards"]
```
